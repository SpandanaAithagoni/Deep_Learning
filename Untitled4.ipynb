```python
import tensorflow as tf
import numpy as np

def causal_mask(size):
    return tf.linalg.band_part(
        tf.ones((size, size)),
        -1,
        0
    )

class PositionalEncoding(tf.keras.layers.Layer):
    def __init__(self, max_len, d_model, **kwargs):
        super().__init__(**kwargs)
        self.max_len = max_len
        self.d_model = d_model
        position = np.arange(max_len)[:, np.newaxis]
        div_term = np.exp(
            np.arange(0, d_model, 2)
            * -(np.log(10000.0) / d_model)
        )
        pe = np.zeros((max_len, d_model))
        pe[:, 0::2] = np.sin(position * div_term)
        pe[:, 1::2] = np.cos(position * div_term)
        self.pe = tf.constant(pe[np.newaxis, ...], dtype=tf.float32)
    def call(self, x):
        seq_len = tf.shape(x)[1]
        return x + self.pe[:, :seq_len, :]
    def get_config(self):
        config = super().get_config()
        config.update({
            "max_len": self.max_len,
            "d_model": self.d_model
        })
        return config

class GPTDecoderBlock(tf.keras.layers.Layer):
    def __init__(self,
                 embed_dim,
                 num_heads,
                 ff_dim,
                 **kwargs):
        super().__init__(**kwargs)
        self.embed_dim = embed_dim
        self.num_heads = num_heads
        self.ff_dim = ff_dim
        self.attention = tf.keras.layers.MultiHeadAttention(
            num_heads=num_heads,
            key_dim=embed_dim
        )
        self.ffn = tf.keras.Sequential([
            tf.keras.layers.Dense(ff_dim, activation="relu"),
            tf.keras.layers.Dense(embed_dim)
        ])
        self.norm1 = tf.keras.layers.LayerNormalization(
            epsilon=1e-6
        )
        self.norm2 = tf.keras.layers.LayerNormalization(
            epsilon=1e-6
        )
    def call(self, x):
        seq_len = tf.shape(x)[1]
        mask = causal_mask(seq_len)
        attn_output = self.attention(
            query=x,
            key=x,
            value=x,
            attention_mask=mask
        )
        out1 = self.norm1(x + attn_output)
        ffn_output = self.ffn(out1)
        return self.norm2(out1 + ffn_output)
    def get_config(self):
        config = super().get_config()
        config.update({
            "embed_dim": self.embed_dim,
            "num_heads": self.num_heads,
            "ff_dim": self.ff_dim
        })

        return config
```

```python
def generate_text(
    seed_text,
    max_tokens=30,
    sequence_length=20
):
    seed_text = seed_text.lower()
    tokens = []
    for word in seed_text.split():
        if word in word2idx:
            tokens.append(word2idx[word])
    if len(tokens) == 0:
        return "No known words found in vocabulary."
    for _ in range(max_tokens):
        input_tokens = tokens[-sequence_length:]
        if len(input_tokens) < sequence_length:
            input_tokens = (
                [0] * (sequence_length - len(input_tokens))
                + input_tokens
            )
        input_array = np.array([input_tokens])
        predictions = model.predict(
            input_array,
            verbose=0
        )
        next_token = np.argmax(
            predictions[0, -1]
        )
        if next_token == 0:
            break
        tokens.append(next_token)
    generated_words = []
    for token in tokens:
        generated_words.append(
            idx2word.get(token, "")
        )
    return " ".join(generated_words)
```

```python
def sample_with_temperature(
    predictions,
    temperature=1.0
):
    predictions = np.asarray(
        predictions
    ).astype("float64")
    predictions = np.log(
        predictions + 1e-8
    ) / temperature
    exp_preds = np.exp(predictions)
    predictions = exp_preds / np.sum(
        exp_preds
    )
    return np.random.choice(
        len(predictions),
        p=predictions
    )
```
