```python
import tensorflow as tf
import numpy as np
import re

from datasets import load_dataset
from tensorflow.keras.layers import (
    Embedding,
    Dense,
    Layer,
    LayerNormalization,
    MultiHeadAttention
)
```

```python
dataset = load_dataset(
    "roneneldan/TinyStories",
    split="train[:1000]"
)

print(dataset[0])
```

```python
texts = [item["text"] for item in dataset]

print(texts[0][:500])
```

```python
cleaned_texts = []
for text in texts:
    text = text.lower()
    text = re.sub(
        r"[^a-zA-Z0-9\s]",
        "",
        text
    )
    cleaned_texts.append(text)
```

```python
tokenized_texts = [
    text.split()
    for text in cleaned_texts
]
```

```python
vocab = sorted(
    set(
        word
        for sentence in tokenized_texts
        for word in sentence
    )
)
word_to_id = {
    word:i
    for i,word in enumerate(vocab)
}
id_to_word = {
    i:word
    for word,i in word_to_id.items()
}
vocab_size = len(vocab)
print("Vocabulary Size:", vocab_size)
```

```python
inputs = []
targets = []
for sentence in tokenized_texts:
    for i in range(1,len(sentence)):
        inputs.append(
            sentence[:i]
        )
        targets.append(
            sentence[i]
        )
```

```python
X = []
for seq in inputs:
    X.append(
        [
            word_to_id[word]
            for word in seq
        ]
    )
y = [
    word_to_id[word]
    for word in targets
]
```

```python
max_len = 20
X = tf.keras.preprocessing.sequence.pad_sequences(
    X,
    maxlen=max_len,
    padding="pre"
)
y = np.array(y)
print(X.shape)
print(y.shape)
```

```python
def create_look_ahead_mask(seq_len):
    return tf.linalg.band_part(
        tf.ones((seq_len,seq_len)),
        -1,
        0
    )
```

```python
class FeedForwardNetwork(Layer):
    def __init__(self,d_model,dff):
        super().__init__()
        self.fc1 = Dense(
            dff,
            activation="relu"
        )
        self.fc2 = Dense(
            d_model
        )
    def call(self,x):
        x = self.fc1(x)
        x = self.fc2(x)
        return x
```

```python
class GPTDecoderBlock(Layer):

    def __init__(
        self,
        d_model,
        num_heads,
        dff
    ):
        super().__init__()
        self.mha = MultiHeadAttention(
            num_heads=num_heads,
            key_dim=d_model//num_heads
        )
        self.ffn = FeedForwardNetwork(
            d_model,
            dff
        )
        self.norm1 = LayerNormalization()
        self.norm2 = LayerNormalization()
    def call(
        self,
        x,
        look_ahead_mask=None
    ):
        attn = self.mha(
            query=x,
            value=x,
            key=x,
            attention_mask=look_ahead_mask
        )
        x = self.norm1(
            x + attn
        )
        ffn_out = self.ffn(x)
        x = self.norm2(
            x + ffn_out
        )
        return x
```

```python
class GPTModel(tf.keras.Model):

    def __init__(
        self,
        vocab_size,
        max_len,
        d_model=128,
        num_heads=4,
        dff=512
    ):

        super().__init__()

        self.token_embedding = Embedding(
            vocab_size,
            d_model
        )

        self.position_embedding = Embedding(
            max_len,
            d_model
        )

        self.decoder1 = GPTDecoderBlock(
            d_model,
            num_heads,
            dff
        )

        self.decoder2 = GPTDecoderBlock(
            d_model,
            num_heads,
            dff
        )

        self.fc = Dense(
            vocab_size
        )

    def call(self,x):

        seq_len = tf.shape(x)[1]

        positions = tf.range(
            seq_len
        )

        token_embed = self.token_embedding(x)

        pos_embed = self.position_embedding(
            positions
        )

        x = token_embed + pos_embed

        mask = create_look_ahead_mask(
            seq_len
        )

        x = self.decoder1(
            x,
            look_ahead_mask=mask
        )

        x = self.decoder2(
            x,
            look_ahead_mask=mask
        )

        last_token = x[:, -1, :]
        return self.fc(last_token)
```

```python
model = GPTModel(
    vocab_size=vocab_size,
    max_len=max_len
)
```

```python
model.compile(
    optimizer="adam",
    loss=tf.keras.losses.SparseCategoricalCrossentropy(
        from_logits=True
    ),
    metrics=["accuracy"]
)
```

```python
y_train = y.reshape(-1,1)
```

```python
def generate_story(
    prompt,
    max_tokens=30
):

    words = prompt.lower().split()

    for _ in range(max_tokens):

        seq = [
            word_to_id[w]
            for w in words
            if w in word_to_id
        ]

        seq = tf.keras.preprocessing.sequence.pad_sequences(
            [seq],
            maxlen=max_len,
            padding="pre"
        )

        logits = model(seq)

        next_token = tf.argmax(
            logits[0],
            axis=-1
        ).numpy()

        next_word = id_to_word.get(
            next_token,
            ""
        )

        words.append(next_word)

    return " ".join(words)
```

```python

story = generate_story(
    "once upon a time"
)

print(story)
```

```python
loss, acc = model.evaluate(
    X,
    y_train
)

perplexity = np.exp(loss)

print("Loss:", loss)
print("Perplexity:", perplexity)
```
