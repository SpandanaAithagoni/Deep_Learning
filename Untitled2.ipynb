```python
texts = [
    "The cat chased the mouse.",
    "The dog barked loudly.",
    "Machine learning is powerful.",
    "Deep learning uses neural networks."
]

print(texts)
```

```python
#CONVERT TO LOWERCASE AND REMOVE UNWANTED SYMBOLS
```

```python
import re
cleaned_texts = []
for text in texts:
    text = text.lower()
    text = re.sub(r'[^a-z\s]', '', text)
    cleaned_texts.append(text)
print(cleaned_texts)
```

```python
#TOKENIZE TEXT
```

```python
tokenized_texts = [text.split() for text in cleaned_texts]
print(tokenized_texts)
```

```python
all_words = []
for sentence in tokenized_texts:
    all_words.extend(sentence)
vocab = sorted(set(all_words))
word_to_idx = {word: idx for idx, word in enumerate(vocab)}
idx_to_word = {idx: word for word, idx in word_to_idx.items()}
print("Vocabulary:")
print(vocab)
print("\nWord to Index:")
print(word_to_idx)
```

```python
import tensorflow as tf
from tensorflow.keras.layers import (
    Layer,
    MultiHeadAttention,
    Dense,
    LayerNormalization,
    Dropout
)
```

```python
#CREATE FEED FORWARD NETWORK
```

```python
class FeedForwardNetwork(Layer):
    def __init__(self, d_model, dff):
        super().__init__()
        self.dense1 = Dense(dff, activation='relu')
        self.dense2 = Dense(d_model)
    def call(self, x):
        x = self.dense1(x)
        x = self.dense2(x)
        return x
```

```python
#CREATE GPT DECODER BLOCK
```

```python
class GPTDecoderBlock(Layer):
    def __init__(self, d_model, num_heads, dff):
        super().__init__()
        self.mha = MultiHeadAttention(
            num_heads=num_heads,
            key_dim=d_model // num_heads
        )
        self.ffn = FeedForwardNetwork(
            d_model,
            dff
        )
        self.norm1 = LayerNormalization()
        self.norm2 = LayerNormalization()
    def call(self, x, look_ahead_mask=None):
        attn_output = self.mha(
            query=x,
            value=x,
            key=x,
            attention_mask=look_ahead_mask
        )
        out1 = self.norm1(
            x + attn_output
        )
        ffn_output = self.ffn(out1)
        out2 = self.norm2(
            out1 + ffn_output
        )
        return out2
```

```python
#DECODER FORWARD PASS
```

```python
def call(self,
             x,
             training=False,
             look_ahead_mask=None):

        # 1. Masked Multi-Head Attention
        attn_output = self.mha(
            query=x,
            value=x,
            key=x,
            attention_mask=look_ahead_mask
        )

        attn_output = self.dropout1(
            attn_output,
            training=training
        )

        # 2. Add & Normalize
        out1 = self.layernorm1(
            x + attn_output
        )

        # 3. Feed Forward Network
        ffn_output = self.ffn(out1)

        ffn_output = self.dropout2(
            ffn_output,
            training=training
        )

        # 4. Add & Normalize
        out2 = self.layernorm2(
            out1 + ffn_output
        )

        return out2
```

```python
def create_look_ahead_mask(seq_len):
    mask = tf.linalg.band_part(
        tf.ones((seq_len, seq_len)),
        -1,
        0
    )
    return mask
```

```python
import tensorflow as tf
from tensorflow.keras.layers import (
    Layer,
    Embedding,
    Dense
)
```

```python
#GPT MODEL CLASS
```

```python
class GPTModel(tf.keras.layers.Layer):

    def __init__(
        self,
        vocab_size,
        max_seq_len,
        d_model,
        num_heads,
        dff
    ):
        super().__init__()

        self.token_embedding = tf.keras.layers.Embedding(
            vocab_size,
            d_model
        )

        self.position_embedding = tf.keras.layers.Embedding(
            max_seq_len,
            d_model
        )

        self.decoder1 = GPTDecoderBlock(
            d_model,
            num_heads,
            dff
        )

        self.decoder2 = GPTDecoderBlock(
            d_model,
            num_heads,
            dff
        )

        self.final_layer = tf.keras.layers.Dense(
            vocab_size
        )

    def call(
        self,
        x,
        look_ahead_mask=None
    ):

        seq_len = tf.shape(x)[1]

        positions = tf.range(
            start=0,
            limit=seq_len,
            delta=1
        )

        token_embed = self.token_embedding(x)

        position_embed = self.position_embedding(
            positions
        )

        x = token_embed + position_embed

        x = self.decoder1(
            x,
            look_ahead_mask=look_ahead_mask
        )

        x = self.decoder2(
            x,
            look_ahead_mask=look_ahead_mask
        )

        logits = self.final_layer(x)

        return logits
```

```python
#CREATE LOOK AHEAD PASS
```

```python
def create_look_ahead_mask(seq_len):

    return tf.linalg.band_part(
        tf.ones((seq_len, seq_len)),
        -1,
        0
    )
```

```python
#CREATE GPT MODEL
```

```python
vocab_size = 100
max_seq_len = 20
d_model = 32
num_heads = 4
dff = 128

gpt = GPTModel(
    vocab_size,
    max_seq_len,
    d_model,
    num_heads,
    dff
)
```

```python
#SAMPLE VOCABULARY
```

```python
vocab = [
    "the",
    "cat",
    "chased",
    "mouse",
    "machine",
    "learning",
    "is",
    "powerful"
]

word_to_id = {
    word: idx
    for idx, word in enumerate(vocab)
}

id_to_word = {
    idx: word
    for word, idx in word_to_id.items()
}
```

```python
#INPUT SENTENCE
```

```python
sentence = [
    "the",
    "cat",
    "chased"
]
token_ids = [
    word_to_id[word]
    for word in sentence
]
print(token_ids)
```

```python
#CONVERT TO TENSOR
```

```python
vocab_size = len(vocab)
gpt = GPTModel(
    vocab_size=vocab_size,
    max_seq_len=20,
    d_model=32,
    num_heads=4,
    dff=128
)
```

```python
input_tokens = tf.constant(
    [token_ids]
)
print(input_tokens)
```

```python
#CREATE MASK
```

```python
mask = create_look_ahead_mask(
    input_tokens.shape[1]
)
```

```python
#MODEL PREDICTION
```

```python
logits = gpt(
    input_tokens,
    look_ahead_mask=mask
)
```

```python
#NEXT TOKEN PROBABILITIES
```

```python
last_token_logits = logits[:, -1, :]
probabilities = tf.nn.softmax(
    last_token_logits,
    axis=-1
)
```

```python
#PREDICT NEXT TOKEN
```

```python
predicted_id = tf.argmax(
    probabilities,
    axis=-1
)
predicted_word = id_to_word[
    predicted_id.numpy()[0]
]
print(
    "Predicted Word:",
    predicted_word
)
```

```python
#GENERATION FUNCTION
```

```python
def generate_text(
    prompt,
    max_new_tokens=5
):
    words = prompt.lower().split()
    for _ in range(max_new_tokens):
        token_ids = [
            word_to_id[word]
            for word in words
        ]
        x = tf.constant([token_ids])
        mask = create_look_ahead_mask(
            len(token_ids)
        )
        logits = gpt(
            x,
            look_ahead_mask=mask
        )
        next_token_logits = logits[:, -1, :]
        probs = tf.nn.softmax(
            next_token_logits,
            axis=-1
        )
        next_id = tf.argmax(
            probs,
            axis=-1
        ).numpy()[0]
        next_word = id_to_word[next_id]
        words.append(next_word)
        print(
            " ".join(words)
        )
```

```python
#GENERATE TEXT
```

```python
generate_text(
    "machine learning",
    max_new_tokens=5
)
```
